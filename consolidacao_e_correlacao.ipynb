Unindo os Dataframes

```python
import pandas as pd

#importando os dados formatados de cada dataframe
df_Inmet = "dados_formatados/FORTALEZA_DADOS_CLIMATICOS.csv"
df_Infodengue = "dados_formatados/INFODENGUE_DADOS_FORTALEZA.CSV"
df_saneamento = "dados_formatados/SANEAMENTO_DADOS_FORTALEZA.csv"

df_Inmet = pd.read_csv(df_Inmet, sep=';', decimal=',', encoding='utf-8-sig')
df_Infodengue = pd.read_csv(df_Infodengue)
df_saneamento = pd.read_csv(df_saneamento)

# 1. Unindo Clima com Dengue
df_final = pd.merge(df_Inmet, df_Infodengue, on='MES_REFERENCIA', how='left')

# 2. Unindo o resultado com Saneamento
df_final = pd.merge(df_final, df_saneamento, on='MES_REFERENCIA', how='left')

# 3. Preenchendo buracos (se houver meses sem registro de dengue)
df_final = df_final.fillna(0)

df_final
```

Padronização de nomes das colunas 

```python
print(df_final.columns)
```

```python
#fazendo alteração dos nome de cada coluna
df_final = df_final.rename(columns={
    'MES_REFERENCIA': 'MESES DE REFERÊNCIA',
    'Temp_Media_Mensal' : 'TEMPERATURA MÉDIA MENSAL (°C)',
    'Umid_Media' : 'UMIDADE MÉDIA (%)',
    'Pres_Media' : 'PRESSÃO MÉDIA (hPa)',
    'casos_mensais' : 'CASOS DE DENGUE (NÚMERO DE CASOS)',
    'nivel_alerta' : 'NÍVEL DE ALERTA - DENGUE (0-3)',
    'Pop_sem_coleta_esgoto' : 'POPULAÇÃO SEM COLETA DE ESGOTO (NÚMERO DE PESSOAS)',
    'Perc_sem_coleta_esgoto' : 'PERCENTUAL SEM COLETA DE ESGOTO (%)'
})

df_final
```

Verificação de periodo

```python
# Verificando o intervalo de datas
print(df_final['MESES DE REFERÊNCIA'].min())
print(df_final['MESES DE REFERÊNCIA'].max())

print("-" * 50)

# Verificando se há valores nulos
print(df_final.isnull().sum())
```

Exportando o dataframe

```python
#Exportando o dataset final para csv
df_final.to_csv('dados_formatados/FORTALEZA_OBS_COMPLETO.csv', sep=';', decimal=',', index=False, encoding='utf-8-sig')
print(f"\nSUCESSO! Total de {len(df_final)} linhas consolidadas em 'FORTALEZA_DADOS_CLIMATICOS_DENGUE_SANEAMENTO.csv'")
```
